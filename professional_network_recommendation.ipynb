### Common Skills Generator

```python
def get_common_skills(skills1, skills2):

    s1 = {
        skill.strip().lower()
        for skill in skills1.split(",")
        if skill.strip()
    }

    s2 = {
        skill.strip().lower()
        for skill in skills2.split(",")
        if skill.strip()
    }

    return sorted(s1.intersection(s2))
```

### Reason Generator

```python
def generate_reason(user1, user2):

    reasons = []

    # Common Skills Match
    common_skills = get_common_skills(
        user1["Skills"],
        user2["Skills"]
    )

    if common_skills:
        reasons.append(
            f"Shared skills: {', '.join(common_skills[:3])}"
        )

    # MBTI Match
    if user1["MBTI"] == user2["MBTI"]:
        reasons.append(
            f"Same MBTI ({user1['MBTI']})"
        )

    # Industry Match
    if user1["Industry"] == user2["Industry"]:
        reasons.append(
            f"Same industry ({user1['Industry']})"
        )

    # Career Goal Match
    if user1["Career_Goal"] == user2["Career_Goal"]:
        reasons.append(
            "Similar career goals"
        )

    # City Match
    if user1["City"] == user2["City"]:
        reasons.append(
            f"Located in {user1['City']}"
        )

    # Experience Match
    Experience_gap = abs(
        user1["Experience_Years"] -
        user2["Experience_Years"]
    )

    if Experience_gap <= 2:
        reasons.append(
            "Similar experience level"
        )

    if len(reasons) == 0:
        return "General profile similarity"

    return "; ".join(reasons)
```

##  Generate Feedback

```python
feedback_records = []

feedback_rng = random.Random(SEED + 2)

for user in users_df["User_ID"]:

    recommendations = re.recommend_hybrid_users(
        user,
        users_df,
        similarity_matrix,
        user_index,
        top_n=5,
        weights={
            "text": 0.50,
            "mbti": 0.20,
            "skills": 0.15,
            "location": 0.10,
            "experience": 0.05
        }
    )

    for _, row in recommendations.iterrows():

        # 1 = accepted, 0 = rejected; deterministic synthetic feedback
        probability = float(row["Compatibility (%)"]) / 100
        feedback = 1 if feedback_rng.random() < probability else 0

        feedback_records.append({
            
            "User_ID": user,
            
            "Recommended_User": row["User_ID"],
            
            "Compatibility": row["Compatibility (%)"],
            
            "Feedback": feedback
        })
```

# Adaptive Learning

```python
# ==========================================
# CREATE UPDATED COMPATIBILITY SCORES
# ==========================================

feedback_df = pd.DataFrame(feedback_records)

def update_feedback_score(score, feedback):

    if int(feedback) == 1:
        return min(score + 5, 100)

    return max(score - 5, 0)


feedback_df["Updated_Compatibility"] = feedback_df.apply(
    lambda row: update_feedback_score(
        row["Compatibility"],
        row["Feedback"]
    ),
    axis=1
)
```

```python
# ==========================================
# GENERATE FEEDBACK TRAINING DATASET
# ==========================================

feedback_records = []
feedback_rng = random.Random(SEED + 1)
BASE_DATE = datetime(2026, 9, 1)

# Generate 5 interactions for every user
for user_id in users_df["User_ID"]:

    source_index = user_index[user_id]

    possible_users = [
        uid for uid in users_df["User_ID"]
        if uid != user_id
    ]

    matched_users = feedback_rng.sample(possible_users, 5)

    for matched_user_id in matched_users:

        target_index = user_index[matched_user_id]

        breakdown = re.calculate_score_breakdown(
            source_index,
            target_index,
            users_df,
            similarity_matrix,
            {
                "text": 0.50,
                "mbti": 0.20,
                "skills": 0.15,
                "location": 0.10,
                "experience": 0.05
            }
        )

        compatibility = breakdown["total_score"]
        probability = compatibility / 100

        timestamp = BASE_DATE - timedelta(
            days=feedback_rng.randint(0, 30),
            hours=feedback_rng.randint(0, 23),
            minutes=feedback_rng.randint(0, 59)
        )

        action = 1 if feedback_rng.random() < probability else 0

        feedback_records.append({
            "user_id": user_id,
            "matched_user_id": matched_user_id,
            "action": action,
            "timestamp": timestamp,
            "text_score": breakdown["text_score"],
            "mbti_score": breakdown["mbti_score"],
            "skill_score": breakdown["skill_score"],
            "location_score": breakdown["location_score"],
            "experience_score": breakdown["experience_score"],
            "compatibility_score": breakdown["total_score"]
        })

feedback_training_df = pd.DataFrame(feedback_records)

feedback_training_df.to_csv(
    "feedback_training.csv",
    index=False
)

print("Feedback Training Dataset Created")
print("Shape:", feedback_training_df.shape)
print("\nAction Distribution:")
print(feedback_training_df["action"].value_counts())
```
